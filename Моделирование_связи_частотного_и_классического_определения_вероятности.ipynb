# Тема 2. Моделирование связи частотного и классического определения вероятности

### Приведите пример, показывающий связь частотного и классического определения вероятности с помощью моделирования (на примере задачи про красно-синие шары или задачи о сумме очков на двух кубиках). Графически показать стремление частотной вероятности к классической при увеличении числа опытов.

## Смоделируем связь частотного и классического определения вероятности на примере европейской рулетки. В рулетке имеется 37 секторов: 18 красных, 18 чёрных и 1 зелёный. Шарик вращается по колесу и останавливается на одном из секторов. Предполагается, что все секторы равновероятны.

## Найти вероятность выпадения красного сектора и проверить результат с помощью компьютерного моделирования.

```python
import numpy as np
import matplotlib.pyplot as plt

# Европейская рулетка:
# 18 красных, 18 черных и 1 зеленый сектор
colors = ['Красное', 'Черное', 'Зеленое']
probabilities = [18/37, 18/37, 1/37]

# Классическая вероятность
p = 18 / 37
print(f'Классическая вероятность: {p:.4f}')

# Частотная вероятность
N = 100000
rng = np.random.default_rng(42)

results = rng.choice(colors, size=N, p=probabilities)
N_A = np.sum(results == 'Красное')

freq = N_A / N
print(f'Частота: {freq:.4f}')

# Сходимость частоты
N_values = np.arange(10, N + 1, 100)
red = (results == 'Красное').astype(int)
cumulative = np.cumsum(red)
P_A_values = cumulative[N_values - 1] / N_values

plt.figure(figsize=(8, 4))
plt.plot(N_values, P_A_values, color='royalblue',
         label='Частота', linewidth=1.5)
plt.axhline(y=p, color='indianred', linestyle='--',
            label='Теоретическая вероятность')
plt.xlabel('N')
plt.ylabel('P(A)')
plt.title('Сходимость частоты выпадения красного сектора')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
