### Пример 1.10

**Условие:** Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово:


```python
letters = 'ТЕИЯРО'
```

#### а) «ТОР»

```python
orders_a = list(permutations(letters, 3))
correct_a = [x for x in orders_a if ''.join(x) == 'ТОР']
P_a = len(correct_a) / len(orders_a)
```

#### б) «ТЕОРИЯ»

```python
orders_b = list(permutations(letters))
correct_b = [x for x in orders_b if ''.join(x) == 'ТЕОРИЯ']
P_b = len(correct_b) / len(orders_b)
```

```python
print("а)", f"{P_a:.4f}")
print("б)", f"{P_b:.4f}")
```

### Пример 1.11

**Условие:** Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

### Пример 1.16

**Условие:** За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что:


```python
people = range(10)

total = 0
alternating = 0
couples = 0

# 0-4 - мужчины, 5-9 - женщины
def same_gender(a, b):
    return (a < 5 and b < 5) or (a >= 5 and b >= 5)

for x in permutations(people):
    total += 1

    if all(not same_gender(x[i], x[(i + 1) % 10]) for i in range(10)):
        alternating += 1

    if all(abs(x.index(i) - x.index(i + 5)) in (1, 9) for i in range(5)):
        couples += 1
```

#### а) Никакие два лица одного пола не сидят рядом

```python
P_a = alternating / total
```

#### б) Супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар

```python
P_b = couples / total
```

```python
print("а)", f"{P_a:.4f}")
print("б)", f"{P_b:.4f}")

```

### Пример 1.17

**Условие:** В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали:

```python
places = list(product(range(9), range(4)))

total = 0
two = 0
seven = 0
three = 0

for x in combinations(places, 7):
    total += 1
    count = len(set(place[0] for place in x))

    if count == 2:
        two += 1
    elif count == 7:
        seven += 1
    elif count == 3:
        three += 1
```

#### а) В два купе


```python
P_a = two / total
```

#### б) В семь купе


```python
P_b = seven / total
```

#### в) В три купе


```python
P_c = three / total
```

```python
print("а)", f"{P_a:.7f}")
print("б)", f"{P_b:.4f}")
print("в)", f"{P_c:.4f}")
```

### Пример 1.43

**Условие**: Наудачу взятый телефонный номер состоит из 5 цифр. Известно, что номер телефона не начинается с цифры ноль.
Какова вероятность того, что в нем все цифры:

```python
digits = '0123456789'

numbers = [x for x in product(digits, repeat=5) if x[0] != '0']
```

#### а) различные

```python
different = [x for x in numbers if len(set(x)) == 5]
P_a = len(different) / len(numbers)
```

#### б) одинаковые

```python
same = [x for x in numbers if len(set(x)) == 1]
P_b = len(same) / len(numbers)
```

#### в) нечетные

```python
odd = [x for x in numbers if all(int(d) % 2 != 0 for d in x)]
P_c = len(odd) / len(numbers)
```

```python
print("а)", round(P_a, 4))
print("б)", round(P_b, 4))
print("в)", round(P_c, 4))
```

### Пример 1.48

**Условие**: В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности:

### Задача про урны

**Условие**: Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

Классическое решение

```python
from math import comb

P_White = 0

for i in range(4):
    p_i = comb(15, i) * comb(5, 3 - i) / comb(20, 3)
    white = (6 + i) / 10
    P_White += p_i * white

print(round(P_White, 4))
```

Моделирование

```python
import random

n = 1000000
count = 0

for _ in range(n):
    urna1 = ["w"] * 15 + ["b"] * 5
    urna2 = ["w"] * 6 + ["b"] * 1

    popped = random.sample(urna1, 3)
    urna2 += popped

    ball = random.choice(urna2)

    if ball == "w":
        count += 1

print(round(count / n, 4))
```

### Пример 2.9

**Условие**: В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.

```python
from scipy.stats import binom, poisson

n = 9
p = 0.2
q = 1 - p
```

#### 1) не будут проданы 5 пакетов

```python
P1 = binom.pmf(4, n, p)
print(round(P1, 4))
```

#### 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов

```python
P2a = binom.cdf(1, n, p)  # менее 2
P2b = binom.cdf(2, n, p)  # не более 2
P2c = binom.sf(1, n, p)   # хотя бы 2
P2g = binom.pmf(1, n, p) + binom.pmf(2, n, p)

print("a)", round(P2a, 4))
print("b)", round(P2b, 4))
print("c)", round(P2c, 4))
print("g)", round(P2g, 4))
```

### Пример 2.10

**Условие**: Завод отправил на базу 10 000 стандартных изделий. Среднее число изделий, повреждаемых при транспортировке, составляет 0,02%. Найти вероятность того, что из 10 000 изделий: 1) будет повреждено: а) 3; б) по крайней мере 3; 2) не будет повреждено: а) 9997; б) хотя бы 9997.

```python
from scipy.stats import binom, poisson

n = 10_000
p = 0.0002
lam = n * p
```

#### 1) будет повреждено: а) 3; б) по крайней мере 3

```python
# Точное биномиальное распределение
P1a = binom.pmf(3, n, p)
P1b = binom.sf(2, n, p)
print("Точные вероятности:")
print("а)", round(P1a, 4))
print("б)", round(P1b, 4))

# Приближение Пуассона
P1a_p = poisson.pmf(3, lam)
P1b_p = poisson.sf(2, lam)
print("\nПриближение Пуассона:")
print("а)", round(P1a_p, 4))
print("б)", round(P1b_p, 4))
```

#### 2) не будет повреждено: а) 9997; б) хотя бы 9997.

```python
# Точное биномиальное распределение
P2a = binom.pmf(3, n, p)
P2b = binom.cdf(3, n, p)
print("Точные вероятности:")
print("а)", round(P2a, 4))
print("б)", round(P2b, 4))

# Приближение Пуассона
P2a_p = poisson.pmf(3, lam)
P2b_p = poisson.cdf(3, lam)
print("\nПриближение Пуассона:")
print("а)", round(P2a_p, 4))
print("б)", round(P2b_p, 4))
```

### Пример 2.11

**Условие**: По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.

```python
from scipy.stats import binom, norm

n = 1000
p = 0.5
q = 1 - p

mu = n * p
sigma = (n * p * q) ** 0.5
```

#### а) 480 предприятий

```python
# Точные вероятности
P_a = binom.pmf(480, n, p)

# Приближение Муавра-Лапласа
P_a_ML = norm.cdf((480.5 - mu) / sigma) - \
         norm.cdf((479.5 - mu) / sigma)

print("а) P(X=480):", P_a)
print("а) Муавр-Лаплас:", P_a_ML)
```

#### б) наивероятнейшее число предприятий

```python
P_b = binom.pmf(500, n, p)

print(round(P_b, 4))
```

#### в) не менее 480

```python
# Точные вероятности
P_v = binom.sf(479, n, p)

# Приближение Муавра-Лапласа
P_v_ML = 1 - norm.cdf((479.5 - mu) / sigma)

print("в) P(X >= 480):", P_v)
print("в) Муавр-Лаплас:", P_v_ML)
```

#### г) от 480 до 520

```python
# Точные вероятности
P_g = binom.cdf(520, n, p) - binom.cdf(479, n, p)

# Приближение Муавра-Лапласа
P_g_ML = norm.cdf((520.5 - mu) / sigma) - \
         norm.cdf((479.5 - mu) / sigma)

print("г) P(480 <= X <= 520):", P_g)
print("г) Муавр-Лаплас:", P_g_ML)
```
